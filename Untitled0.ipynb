```python
#create a function to run queries
def run(query):
  return pd.read_sql_query(query,conn)
```

answer ✍️ 5 question ❓ using SQL

```python
#Q1:How much is each member borrowing?
query='''SELECT
      m.first_name || ' ' || m.last_name AS member_name,
       COUNT(c.checkout_id) AS total_checkouts
       FROM members m
       LEFT JOIN checkouts c
       ON m.member_id = c.member_id
      GROUP BY
      m.first_name,
      m.last_name
    '''
print("each student number of books borrowed: ")
run(query)
```

```python
#Q2:Which books match a chosen author pattern?
query='''
SELECT title,author FROM books
WHERE author LIKE '%ou%'
'''
print("books contain 'ou' : ")
run(query)
```

```python
#Q3:What are the most popular books?"FRIST 5"
query='''SELECT b.title ,
COUNT(c.checkout_id) AS total_checkouts
FROM books b
LEFT JOIN checkouts c ON b.book_id = c.book_id
GROUP BY b.title
ORDER BY
total_checkouts DESC
LIMIT 5'''
print("top 5 borrowed books:")
run(query)
```

```python
#Q4:4. Who are the most active readers?
query='''SELECT m.first_name || ' ' || m.last_name AS member_name,
         COUNT(c.checkout_id) AS BOROWS
         FROM members m
         LEFT JOIN checkouts c
         ON m.member_id = c.member_id
         GROUP BY
         m.first_name,
         m.last_name
         ORDER BY
         BOROWS DESC
         LIMIT 10'''
print("top 10 readers:")
run(query)
```

```python
#Q5:What does a neighborhood's activity look like further back in time?
query='''SELECT
m.first_name || ' ' || m.last_name AS member_name,
         COUNT(c.checkout_id) AS BOROWS
         FROM members m
         JOIN checkouts c
         ON m.member_id = c.member_id
         WHERE m.neighborhood ='Maadi'
         GROUP BY
         m.member_id, m.first_name, m.last_name
         ORDER BY
         c.checkout_date DESC
         limit 10
         OFFSET 10
         '''
print("readers actvity in Maadi neighborhood:")
run(query)
```

merge the 3 sources file

```python
#merge checkouts and members dataframes 
stage1 = checkouts.merge(
    members,
    on="member_id",
    how="left")
print(stage1.shape)
stage1.head()
```

```python
#merge stage1(members and checkouts) and books dataframes
stage2 = stage1.merge(
    books,
    on="book_id",
    how="left")
print(stage2.shape)
stage2.head()
```

```python
#read json file 
Book_Catalog = pd.read_json("Book Catalog.json")
```

```python
#explore the json data
Book_Catalog.head()
```

```python
#merge stage2(members, checkouts, and books) and Book_Catalog 
#to add each book information
stage3 = stage2.merge(
    Book_Catalog,
    on="book_id",
    how="left")
print(stage3.shape)
stage3.head()
```

```python
Reading_Kickoff = pd.read_html("Reading Kickoff Signups.html")

reading_kickoff = Reading_Kickoff[0]#to take the table
print(reading_kickoff.shape)
```

```python
#explore the reading_kickoff data 
reading_kickoff.head()
```

```python
#change coulmn names from capital to small to be able to be added
reading_kickoff = reading_kickoff.rename(columns={
    'Member ID': 'member_id',
    'Book ID': 'book_id',
    'Checkout Date': 'checkout_date'
})

```

```python
#ensure that the column names are changed
reading_kickoff.head()
```

```python
#adjust the reading_kickoff data to be able to merge with other data
kickoff_stage1 = reading_kickoff.merge(members, on='member_id', how='left')
kickoff_stage2 = kickoff_stage1.merge(books, on='book_id', how='left')
kickoff_stage3 = kickoff_stage2.merge(Book_Catalog, on='book_id', how='left')
```

```python
#merge all data in one dataset
final_dataset = pd.concat([stage3, kickoff_stage3], ignore_index=True)
final_dataset.shape
```
